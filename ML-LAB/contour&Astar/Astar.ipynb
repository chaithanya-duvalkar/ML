```python
# Write a program to implement the A* algorithm 

from queue import PriorityQueue

def astar(graph,h,start,goal):
    open_set=PriorityQueue()
    open_set.put((0,start))
    came_from={}
    g_score={node:float ('inf') for node in graph}
    g_score[start]=0

    while not open_set.empty():
        _,current=open_set.get()

        if current==goal:
            path=[]
            total_score=g_score[goal]
            while current in came_from:
                path.append(current)
                current=came_from[current]
            path.append(start)
            path.reverse()
            print(f"path found:{path},cost:{total_score}")
            return path,total_score
        for neighbors,cost in graph[current]:
            tentative_g=g_score[current]+cost
            if tentative_g<g_score[neighbors]:
                came_from[neighbors]=current
                g_score[neighbors]=tentative_g
                f_score=tentative_g+h[neighbors]
                open_set.put((f_score,neighbors))
    print("no path found")
    return None
graph={
    'S':[('A',1),('G',10)],
    'A':[('B',2),('C',1)],
    'B':[('D',5)],
    'C':[('D',3),('G',4)],
    'D':[('G',2)],
    'G':[]
}                
h={
    'S':5,
    'A':3,
    'B':4,
    'C':2,
    'D':6,
    'G':0
}
astar(graph,h,'S','G')

        
```
